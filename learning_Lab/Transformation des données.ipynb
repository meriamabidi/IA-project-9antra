# codage des variables catégorielles

```python
etat_mapping = {
    'Inprogress': 1,
    'Terminated': 0,
    # Ajoutez autant de cas que nécessaire pour chaque valeur de l'énumération
}

events['status_event_encoded'] = events['Status_event'].map(etat_mapping)

# Afficher le DataFrame avec la colonne "etat_quantitatif"
print(events)
```

```python
import pandas as pd
from datetime import datetime

# Assuming 'date_event' is a column in a pandas DataFrame
events['date_event'] = events['date_event']
# Dictionnaire pour mapper les noms de mois abrégés en français aux mois numériques
mois_mapping = {
    'JANV': 1,
    'FÉVR': 2,
    'MARS': 3,
    'AVR': 4,
    'MAI': 5,
    'JUIN': 6,
    'JUIL': 7,
    'AOÛT': 8,
    'SEPT': 9,
    'OCT': 10,
    'NOV': 11,
    'DÉC': 12
}

# Function to format the date
def format_date(date_string):
    # Extraction du jour, du mois et de l'année
    _, jour, mois_abrege, annee = date_string.split()

    # Conversion du mois abrégé en numéro de mois
    mois = mois_mapping.get(mois_abrege.rstrip('.'), 0)

    # Checking if the month is valid
    if mois:
        # Création de l'objet de date
        date_obj = datetime(int(annee), mois, int(jour.rstrip(',')))

        # Formatting the date as "dd/mm/yyyy"
        formatted_date = date_obj.strftime("%d/%m/%Y")
        return formatted_date
    else:
        return f"Invalid month abbreviation: {mois_abrege}"

# Apply the formatting function to create a new column 'new_date'
events['new_date_event'] = events['date_event'].apply(format_date)

# Display the DataFrame with the new column
print(events)
```

```python
# Function to convert numbers ending with "k" to numeric values
def convert_to_numeric(value):
    try:
        if isinstance(value, str) and 'K' in value:
        # Replace 'K' or 'k', remove commas, and convert to float
            return float(str(value).replace('K', '').replace('k', '').replace(',', '')) * 1000
        elif isinstance(value, (float, int)):
            return float(value)
        else:
            return None  # Handle other cases if necessary
    except ValueError:
        return None  # Handle other cases if necessary

# Apply the function to the "participants" and "interested" columns
events['participants'] = events['participants'].apply(convert_to_numeric)
events['interested'] = events['interested'].apply(convert_to_numeric)

# Display the DataFrame with numeric values
print("\nDataFrame with numeric values:")
print(events)

```

# mise à l échelle des variables numériques

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Read the file into dataframes
import pandas as pd
data = pd.read_excel("DATA BI.xlsx",index_col=False,keep_default_na=True,sheet_name='DEPENSES', header=2)


# Filtrer les colonnes numériques
numeric_columns = data.select_dtypes(include=['float64', 'int64']).columns

# Créer un MinMaxScaler
min_max_scaler = MinMaxScaler()

# Appliquer la normalisation aux colonnes numériques uniquement
data_normalized = data.copy()
data_normalized[numeric_columns] = min_max_scaler.fit_transform(data_normalized[numeric_columns])

# Créer un StandardScaler
standard_scaler = StandardScaler()

# Appliquer la standardisation aux colonnes numériques uniquement
data_standardized = data.copy()
data_standardized[numeric_columns] = standard_scaler.fit_transform(data_standardized[numeric_columns])

# Afficher les données normalisées et standardisées
print("Données normalisées :")
print(data_normalized)

print("\nDonnées standardisées :")
print(data_standardized)
```
